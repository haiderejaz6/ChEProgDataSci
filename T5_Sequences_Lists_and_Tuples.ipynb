# 1. Creating and Indexing Lists

### Concept
A list holds an ordered sequence of values in square brackets. Index 0 is the first item; negative indices count back from the end, so -1 is the last.

*Ties to T-03.01.03 Pressure, Temperature and Density (KA-03)*

```python
T_K = [350.0, 352.0, 351.0, 355.0, 358.0]   # reactor readings, K

print(T_K[0])          # first reading
print(T_K[-1])         # last reading
print(len(T_K))        # how many readings
print(T_K[1] + T_K[2]) # elements work in expressions
```

Detailed notes: A list is a sequence: its elements have positions, starting at 0. For a list of length n, valid positive indices run from 0 to n - 1, and valid negative indices from -1 to -n; T_K[-1] is shorthand for T_K[len(T_K) - 1]. Anything outside that range raises IndexError. Lists usually hold values of one kind (all temperatures), though Python allows mixed types. The zero-based convention trips up students who count from 1 on paper; have them say the index out loud when reading code, for example T_K[2] is the third reading.

### Activity: predict then run
Using `T_K` from above (5 readings), what does each line print?

```python
print(T_K[-2])
print(T_K[len(T_K) - 1])
print(T_K[5])
```

Answer: 355.0, then 358.0 (the last element, same as T_K[-1]), then IndexError: list index out of range. With 5 items the last valid positive index is 4.

# 2. Mutability and Growing a List

### Concept
Lists are mutable: you can change an element in place, or add to the end with `append` or `+=`. Two lists join with `+` into a new list.

```python
flows = [12.5, 8.0]           # stream flow rates, mol/s
flows[1] = 8.4                # correct a reading in place
flows.append(4.0)             # add one item to the end
flows += [6.2, 3.1]           # extend by another list
print(flows)
print(flows + [0.5])          # + builds a new list
```

Detailed notes: Mutability means the same list object changes; no new list is created by item assignment, append, or +=. The + operator, by contrast, builds a brand-new list and leaves both operands untouched. += on a list behaves like extend: the right-hand side must be iterable, and each of its elements is added separately. That detail is the bug in the activity. Strings and tuples are also sequences but are immutable, so item assignment on them raises TypeError.

### Activity: spot the bug
This should add one more component name to the list. It runs, but the list is wrong. What happened?

```python
components = ["Benzene", "Toluene"]
components += "Xylene"
print(components)
```

Answer: prints ['Benzene', 'Toluene', 'X', 'y', 'l', 'e', 'n', 'e']. += iterates over its right-hand side, and a string is a sequence of characters, so each letter is appended separately. Fix: components.append('Xylene') or components += ['Xylene'].

# 3. Tuples, Unpacking, and enumerate

### Concept
A tuple `( )` is an immutable sequence, a good fit for one fixed record. Unpacking assigns its items to names. `enumerate` gives each item with its index.

*Ties to T-03.01.03 Pressure, Temperature and Density (KA-03)*

```python
feed = ("Feed-1", 350.0, 5.0)        # (name, T in K, P in bar)
name, T, P = feed                     # unpack the record
print(f"{name}: {T} K, {P} bar")

readings = [350.0, 352.0, 351.0]
for i, T in enumerate(readings, start=1):
    print(f"Reading {i}: {T} K")
```

Detailed notes: The usual convention is that a list holds many things of the same kind (all the readings) while a tuple holds one thing made of different parts (a stream's name, temperature, and pressure). Tuples cannot be changed after creation, which makes them safe to pass around and usable as dictionary keys in T6. Unpacking needs exactly as many names as items or it raises ValueError. enumerate pairs each item with a counter, replacing the error-prone range(len(...)) pattern; start=1 shifts the counter for human-readable numbering. It is the parentheses plus the comma that make a tuple, and the comma is what matters.

### Activity: predict then run
Both lines look like one-element tuples. What type does each print?

```python
a = ("Feed-1")
b = ("Feed-1",)
print(type(a), type(b))
```

Answer: <class 'str'> <class 'tuple'>. Parentheses alone only group an expression; the trailing comma is what makes a one-element tuple.

# 4. Slicing

### Concept
`seq[start:end:step]` returns a new sequence from `start` up to, but not including, `end`. Leaving a part out means from the beginning, to the end, or a step of 1.

```python
T = [350, 352, 351, 355, 358, 360, 359, 361]   # K, one per minute

print(T[2:5])      # minutes 2, 3, 4
print(T[:3])       # first three
print(T[-3:])      # last three
print(T[::2])      # every second reading
print(T[::-1])     # reversed copy
```

Detailed notes: The end index is exclusive, which is what makes T[:3] + T[3:] rebuild the original list and makes len(T[a:b]) equal b - a. A slice always returns a new list (a shallow copy), so T[:] is a quick way to copy a list. Out-of-range slice bounds do not raise errors; they are clipped. Slices can also appear on the left of an assignment to replace or delete a range (T[0:2] = [] deletes two items), and del T[0:2] does the same. NumPy arrays in T7 use the same syntax, with one important difference covered there: array slices are views, not copies.

### Activity: predict then run
Using the same `T`, what does each slice print?

```python
print(T[1:6:2])
print(T[5:100])
print(T[-1:-4:-1])
```

Answer: [352, 355, 360] (indices 1, 3, 5), then [360, 359, 361] (the end is clipped, no error), then [361, 359, 360] (walking backwards from the last element, stopping before index -4).

# 5. Sorting and Searching

### Concept
`list.sort()` sorts a list in place and returns None. `sorted()` returns a new sorted list. `in` tests membership; `index` finds a position.

```python
bp_C = [110.6, 80.1, 138.4]                   # boiling points, C
print(sorted(bp_C))                           # new list, original kept
bp_C.sort(reverse=True)                       # sorts in place
print(bp_C)

names = ["Toluene", "Benzene", "Xylene"]
print("Benzene" in names, names.index("Xylene"))
print(any(t > 130 for t in bp_C), all(t > 50 for t in bp_C))
```

Detailed notes: Methods that change a list in place (sort, append, reverse) return None by design, so their result should never be assigned. sorted works on any iterable, including tuples and strings, and always returns a list. Both accept reverse=True and a key function (key=len, or a lambda from T4). index raises ValueError when the item is missing, so check with in first. any and all reduce a sequence of True/False values to a single answer, a compact way to ask whether any reading is over a limit or whether all readings are in range.

Answer: prints None. sort changes temps in place and returns None, so ranked is None. Fix: ranked = sorted(temps), or call temps.sort() on its own line and print temps.

# 6. More List Methods

### Concept
`insert`, `extend`, `remove`, `pop`, `count`, `reverse`, and `copy` cover most everyday edits to a list.

```python
units = ["pump", "reactor", "column"]
units.insert(1, "heater")          # insert at index 1
units.extend(["condenser"])        # add several items
units.remove("pump")               # remove first match by value
last = units.pop()                 # remove and return the last item
print(units, last)
backup = units.copy()              # an independent copy
```

Detailed notes: remove deletes by value (the first match only) and raises ValueError if absent; pop deletes by index, defaulting to the last item, and returns it, which is how a list acts as a stack (push with append, pop from the end). insert shifts everything after the index to the right, which is slow for very long lists. copy (or list(x), or x[:]) makes a new outer list, so changes to the copy's structure do not affect the original. Plain assignment, backup = units, creates no copy at all, just a second name for the same list, as T4 showed.

### Activity: trace the variable
Track `line` after each statement. What is printed?

```python
line = ["P-101", "E-101", "R-101"]
line.append("T-101")
line.insert(0, "V-101")
line.pop(2)
print(line)
```

Answer: ['V-101', 'P-101', 'R-101', 'T-101']. After append: [P, E, R, T]; after insert at 0: [V, P, E, R, T]; pop(2) removes index 2, which is now E-101, not R-101. The shift caused by insert is the point.

# 7. List Comprehensions and Generator Expressions

### Concept
`[expr for x in seq if cond]` builds a new list in one line: it maps (transforms) and optionally filters. With `( )` instead of `[ ]` it becomes a lazy generator.

*Ties to T-03.01.02 Composition Measures (KA-03)*

```python
masses_kg = [40.0, 10.0, 30.0, 20.0]           # component masses
total = sum(masses_kg)

w = [m / total for m in masses_kg]             # mass fractions
print(w)
major = [x for x in w if x >= 0.25]            # filter
print(major)
print(sum(m ** 2 for m in masses_kg))          # generator, no list
```

Detailed notes: A comprehension replaces the four-line pattern of creating an empty list, looping, optionally testing, and appending. It is the preferred Python spelling of map and filter from T4. A generator expression produces items one at a time on demand (lazy evaluation) instead of building the whole list first (greedy evaluation); passed straight to sum, min, max, any, or all, it saves memory, which matters for very large data. Readability rule: if a comprehension needs more than one if or a nested loop, a plain for loop is usually clearer.

### Activity: predict then run
Mass fractions from above are 0.4, 0.1, 0.3, 0.2. What does this print?

```python
labels = ["A", "B", "C", "D"]
print([labels[i] for i in range(len(w)) if w[i] < 0.25])
print(len([x for x in w if x > 1]))
```

Answer: ['B', 'D'], then 0. The filter keeps the indices whose fraction is below 0.25; no fraction exceeds 1, so the second list is empty. A comprehension with nothing passing the filter gives [], not an error.

# 8. zip and Key Functions

### Concept
`zip` walks several sequences side by side, pairing items by position. `min`, `max`, and `sorted` accept `key=` to decide what to compare.

```python
names = ["Benzene", "Toluene", "Xylene"]
bp_C = [80.1, 110.6, 138.4]            # normal boiling points, C

for n, bp in zip(names, bp_C):
    print(f"{n:<8} {bp:>6.1f} C")

pairs = list(zip(names, bp_C))
print(max(pairs, key=lambda p: p[1]))  # compare by boiling point
```

Detailed notes: zip stops at the shortest input, silently, so mismatched lengths lose data without an error; worth checking len first when the lists come from different sources. Each zipped item is a tuple, which is why unpacking in the for line works. Without key, max compares tuples element by element starting with the first, so it would pick the alphabetically last name; key=lambda p: p[1] tells it to compare by the second element. reversed(seq) is the other common helper: it iterates backwards without copying.

### Activity: predict then run
What do these two lines print?

```python
print(min(pairs, key=lambda p: p[1])[0])
print(list(zip(names, [1.0, 2.0])))
```

Answer: Benzene (lowest boiling point, then [0] takes the name), then [('Benzene', 1.0), ('Toluene', 2.0)]. Xylene disappears because zip stops at the shorter list.

# 9. Two-Dimensional Lists

### Concept
A list of lists acts as a table: `table[row][col]`. Nested loops walk every cell, row by row.

```python
# rows: streams S1..S3, columns: water, ethanol, methanol (kg/h)
flows = [[70.0, 25.0, 5.0],
         [20.0, 75.0, 5.0],
         [50.0, 40.0, 10.0]]

print(flows[1][0])                    # stream S2, water
for row in flows:
    print(sum(row))                   # total flow of each stream
```

Detailed notes: The outer list holds rows; each row is itself a list, so flows[1] is the whole second row and flows[1][0] its first entry. Rows are easy to reach; columns are not, since a column has to be gathered from every row with a loop or comprehension ([row[1] for row in flows]). That asymmetry is one of the main reasons NumPy arrays in T7 exist: flows[:, 1] selects a column directly. Warn students against building a table with [[0] * 3] * 3, which repeats one inner list three times, so changing one cell changes a whole column.

### Activity: predict then run
Using `flows` above, what does each line print?

```python
print(flows[2][1])
print([row[1] for row in flows])
print(len(flows), len(flows[0]))
```

Answer: 40.0 (stream S3, ethanol); [25.0, 75.0, 40.0] (the ethanol column); 3 3 (three rows, three columns).

# 10. Putting It Together: A Stream Composition Table

### Concept
Combine a 2D list, a comprehension, zip, and f-string formatting to turn raw component flows into a readable composition report.

*Ties to T-03.01.02 Composition Measures (KA-03)*

```python
streams = ["S1", "S2", "S3"]
comps = ["water", "ethanol", "methanol"]
flows = [[70.0, 25.0, 5.0], [20.0, 75.0, 5.0], [50.0, 40.0, 10.0]]

for name, row in zip(streams, flows):
    total = sum(row)
    w = [m / total for m in row]               # mass fractions
    main = comps[w.index(max(w))]              # dominant component
    print(f"{name}: {total:5.1f} kg/h, mostly {main} ({max(w):.2f})")
```

Detailed notes: No new syntax here; this is a synthesis of the lecture. zip pairs each stream name with its row, sum totals the row, a comprehension turns flows into fractions, max and index find the dominant component, and an f-string with field widths lines the report up. This is the same shape as the T3 monitoring loop (iterate, transform, summarize), now operating on a table instead of a flat list. In T6 the same data become a dictionary keyed by component name, which removes the need to keep comps and the column order in sync.

### Activity: cold call
If S3's methanol flow rose from 10.0 to 60.0 kg/h, what would the S3 line print as the dominant component, and why?

```python
flows[2][2] = 60.0
row = flows[2]
w = [m / sum(row) for m in row]
print(comps[w.index(max(w))], round(max(w), 2))
```

Answer: methanol 0.4: the S3 total becomes 150 kg/h, so methanol (60) edges out water (50, fraction 0.33). Look for students who reason from the raw flows rather than the fractions; either works here, since the total is shared.
