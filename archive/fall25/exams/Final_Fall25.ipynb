## Section C — Interpreting Python code for climate data analysis (10 marks)

**Task:** Carefully examine the Python script below. For each logically separated block of code,
interpret its purpose in the overall workflow. In your own words, explain what each section is
doing and why it is necessary for conducting a climate-trend analysis.

This one is read-and-explain, not run-and-see: the script expects `nyc_january_avg_highs.csv`,
which was not part of the exam. Write your explanation block by block, then — if you want to see
it work — point `read_csv` at the temperature data from the T10 notebook.

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import linregress


df = pd.read_csv('nyc_january_avg_highs.csv', skiprows=2)
df.head()


df.columns = ['Date', 'Temperature']


df.info()

# Extract year from the 'Date' column (first 4 digits)
df["Year"] = df["Date"].astype(str).str[:4].astype(int)


df = df[["Year", "Temperature"]]
df.head()


sns.set_style("whitegrid")
plt.figure(figsize=(5,3))
sns.scatterplot(data=df, x="Year", y="Temperature")
plt.title("NYC Average January High Temperatures (1895-2025)")
plt.xlabel("Year")
plt.ylabel("Temperature (°F)")
plt.show()


slope, intercept, r_value, p_value, std_err = linregress(df["Year"], df["Temperature"])
print(f"Slope (m): {slope:.4f}")
print(f"Intercept (b): {intercept:.2f}")
print(f"R-squared: {r_value**2:.3f}")


plt.figure(figsize=(5,3))
sns.regplot(data=df, x="Year", y="Temperature", line_kws={"color":"red"})
plt.title("Trend of NYC January High Temperatures (1895-2025)")
plt.xlabel("Year")
plt.ylabel("Temperature (°F)")
plt.show()


year = 2026
predicted_temp = slope * year + intercept
print(f"Predicted January Avg High in {year}: {predicted_temp:.2f} °F")


pd.set_option('display.precision', 2)
df['Temperature'].describe()
````
