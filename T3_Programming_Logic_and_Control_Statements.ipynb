# 1. Algorithms and Pseudocode

### Concept
An algorithm is a step-by-step procedure to solve a problem. Pseudocode writes those steps in plain English before any Python is written, so logic errors surface early.

```python
mass_A = float(input("Mass of A, kg: "))   # mass of component A, kg
mass_B = float(input("Mass of B, kg: "))   # mass of component B, kg

total = mass_A + mass_B    # total mixture mass, kg
x_A = mass_A / total       # mass fraction of A (0 to 1)
print(f"x_A = {x_A:.2f}")
```

Detailed notes: An algorithm only needs to be unambiguous and finite; it is not written in any particular language. Pseudocode is one convenient plain-English notation for it, not a formal standard. Writing INPUT/COMPUTE/DISPLAY style pseudocode before code forces you to separate what has to happen from how Python's syntax expresses it, which is exactly the debugging skill this course exercises later: when a program misbehaves, restate it as pseudocode first to isolate whether the logic or the syntax is wrong. This mirrors process design, where a block flow diagram is the equivalent sketch before choosing equipment and controls.

### Activity: cold call
The pseudocode was: INPUT mass_A, mass_B; COMPUTE total; COMPUTE x_A; DISPLAY x_A. Which line above matches COMPUTE x_A?

Answer: `x_A = mass_A / total`. Walking students through matching each pseudocode line to its Python line checks whether they can read structure, not just syntax.

# 2. Flowcharts

### Concept
A flowchart draws the same logic as a picture: an oval for start/end, a rectangle for a step, a diamond for a decision, arrows for flow. The same shapes extend to elif chains and loops.

### Example
![If-else flowchart](images/if_else_flowchart.jpg)

Detailed notes: A flowchart's shapes are a light convention, not a strict standard: oval for start/end, rectangle for a processing step, diamond for a yes/no decision, arrows for control flow. What matters pedagogically is that every diamond has exactly two exits (true/false), the same constraint an if/else block has in code; a diamond drawn with three arrows out is usually a student trying to skip writing an elif. Flowcharts are most useful for planning nested decisions and loops before typing anything, and they read directly onto the if-elif-else and while/for constructs covered later in this lecture.

Answer (one valid version): diamond 1, "T < 0 C?" (yes -> solid); diamond 2, "T < 100 C?" (yes -> liquid, no -> gas). Listen for students who get the boundary conditions backwards.

# 3. The if Statement

### Concept
An `if` block runs only when its condition is True. Indentation, not braces, marks the block; inconsistent indentation is a common source of errors.

```python
pressure_bar = 6.0        # vessel pressure reading, bar
if pressure_bar > 5.0:    # design limit for this vessel, bar
    print("Over design pressure")
```

Detailed notes: Python has no braces or end keywords marking a block, so indentation is the block. The interpreter tracks the current indentation level, and any change that doesn't match an existing level raises an IndentationError. This is why mixing tabs and spaces, or copy-pasting code from a source with a different indent width, is a classic source of bugs; have students set their editor to insert 4 spaces per tab and make whitespace visible. Also worth noting: an if with a False condition simply skips the block silently, there is no implicit else unless one is written, which trips up students expecting C-style behavior.

### Activity: spot the bug
This should print both lines whenever the grade passes. What is actually wrong with it?

```python
grade = 85                  # student's exam score, out of 100
if grade >= 60:              # passing threshold
    print("Passed")
  print("Good job")
```

Answer: raises IndentationError. The second print is meant to run after the if block, not inside it, but its indentation does not match any valid level. Consistent 4-space indentation avoids this.

# 4. if-elif-else and Boolean Operators

### Concept
`if-elif-else` picks exactly one branch from several, checked top to bottom. `and` / `or` combine conditions into one.

```python
flow_mol_s = 45.0          # molar flow rate, mol/s
if flow_mol_s < 10:
    rate = "low"
elif flow_mol_s < 50:
    rate = "medium"
else:
    rate = "high"
print(f"Flow rate: {rate}")
```

Detailed notes: elif chains are evaluated top to bottom and stop at the first True condition, so ordering matters; a common bug is putting a broad condition before a narrower one that can then never be reached. and/or short-circuit: in 'a and b', b is never evaluated if a is False, which matters when a later condition depends on an earlier one being true first, such as guarding a division with '(b != 0) and (a / b > 1)'. The Reynolds number example previews Fluid Mechanics (CHE-214): the same if/else shape reappears there classifying flow regime, later with a transitional band instead of one sharp cutoff.

**Sneak Peek: T-05.02.03 Laminar/Turbulent Flow (KA-05)**
The same if-elif-else shape classifies flow regime by Reynolds number. You will study this fully in CHE-214, Fluid Mechanics.

### Activity: predict then run
What does this print for `Re = 1800`?

```python
Re = 1800          # Reynolds number: inertial vs. viscous forces
if Re < 2300:       # laminar-turbulent transition, pipe flow
    regime = "laminar"
else:
    regime = "turbulent"
print(regime)
```

Answer: laminar, since 1800 is below the 2300 cutoff. Try Re=4000 live afterward to show the other branch (turbulent).

# 5. The while Statement

### Concept
A `while` loop repeats as long as its condition stays True. Something inside the loop must eventually make the condition False, or it never stops.

```python
volume_L = 100.0        # liquid volume in the tank, L
while volume_L > 20.0:  # stop once at/below the low-level setpoint, L
    volume_L -= 15.0    # each cycle drains 15 L
    print(f"Remaining: {volume_L:.0f} L")
```

Detailed notes: A while loop re-checks its condition before every iteration, including the first one, so a False condition up front means the loop body never runs at all (unlike a do-while in other languages). The loop is only as good as its exit condition: something the loop body actually changes must appear in that condition, or it degenerates into an infinite loop, the single most common while-loop bug students introduce. When tracing a while loop by hand, ask what has to change, and does this line change it, before running anything.

### Activity: spot the bug (do not run this cell live)
This should stop once the level drops below 5 m, but it hangs forever if run. Why?

```python
level_m = 8.0          # tank level, m
while level_m > 5.0:    # stop at the low-level mark, m
    print(f"Level: {level_m} m")
    level_m = 8.0
```

Answer: level_m is reset to 8.0 every iteration instead of decreasing, so `level_m > 5.0` stays True forever. Reason through this one out loud rather than running it; show the fix (`level_m -= ...`) immediately after.

# 6. The for Statement

### Concept
A `for` loop repeats once per item in a sequence, no manual counter or exit condition needed.

```python
streams = ["Feed-1", "Feed-2", "Product-1"]   # streams to process
for s in streams:                    # s = current stream name
    print(f"Processing {s}")
```

Detailed notes: A for loop in Python iterates directly over the items of a sequence (list, string, range, etc.); there is no manual index bookkeeping unless enumerate() is used on purpose. This removes an entire class of off-by-one bugs common in counted loops written with an explicit index. The terminal-velocity example is a pattern students will reuse constantly: take one formula and apply it once per item in a list, which is also the conceptual bridge to vectorized NumPy operations later in the course, where the same computation runs with no explicit loop.

**Sneak Peek: T-09.02.01 Terminal Velocity, Stokes' Law (KA-09)**
The same for loop can sweep across particle sizes. You will study particle settling fully in CHE-346, Particulate Technology.

### Activity: predict then run
Stokes' law gives terminal velocity for small particles. What does this print for each size?

```python
sizes_um = [10, 50, 100]   # particle diameters, micrometers
for d in sizes_um:
    # Stokes' law: v ~ d^2 * rho_p * g / (18*mu); mu = water viscosity
    v = (d * 1e-6) ** 2 * 1600 * 9.81 / (18 * 0.001)   # velocity, m/s
    print(f"{d} um -> {v:.2e} m/s")
```

Answer: velocity grows with the square of particle size, so the 100 um particle settles about 100 times faster than the 10 um one. Exact values matter less than seeing one formula reused once per list item.

# 7. Nested Loops, break, and continue

### Concept
A loop inside another loop runs the inner loop fully for every step of the outer one. `break` exits a loop early; `continue` skips straight to the next iteration.

```python
streams = ["Feed-1", "Feed-2"]        # outer loop: stream names
components = ["Water", "Ethanol"]      # inner loop: component names
for s in streams:                       # s = current stream
    for c in components:                # c = current component
        print(f"{s}: {c}")
```

Detailed notes: In a nested loop, the inner loop completes all of its iterations for each single iteration of the outer loop, so total iterations = len(outer) * len(inner); this multiplies quickly and is worth a quick aside if students ask why a triple-nested loop over large lists runs slowly. break exits only the innermost loop it's written in, not all enclosing loops, a frequent misconception; escaping multiple levels needs a flag variable or restructuring into a function with a return. continue skips the rest of the current iteration's body, but the loop's own condition check still happens as normal on the next pass.

**Sneak Peek: T-15.01.03 Friction (KA-15)**
A loop over contact surfaces summing friction force follows the same one-formula-per-item shape. You will study this fully in a later mechanics or particulate-systems course.

### Activity: predict then run
`F = mu * N`. What does this print for each normal force?

```python
N_values = [50, 80, 120]   # normal force on each surface, N
mu = 0.3                   # coefficient of friction (dimensionless)
for N in N_values:
    F = mu * N              # friction force: F = mu * N, N
    print(f"N={N} N -> friction={F:.1f} N")
```

Answer: 15.0, 24.0, 36.0 N. Friction scales linearly with normal force here. Worth pointing out this loop has the exact same shape as the terminal-velocity sneak peek two sections ago, one formula, applied once per list item.

# 8. Augmented Assignments

### Concept
`x += 5` means the same as `x = x + 5`. Shorthand operators (`+= -= *= /=`) are common inside loops that accumulate a running value.

```python
total_flow = 0.0     # running total flow rate, mol/s
total_flow += 12.5   # add flow from stream 1, mol/s
total_flow += 8.0    # add flow from stream 2, mol/s
print(f"Total flow = {total_flow} mol/s")
```

Detailed notes: For immutable types (numbers, strings, tuples), x += 5 is equivalent to x = x + 5. For mutable objects such as lists, += can mutate the object in place instead of creating a new one, which matters if another variable references the same list. A common beginner error is writing total += reading before total has ever been initialized with total = 0.0; += requires the name to already exist and hold a compatible type.

### Activity: predict then run
What does `x` equal after these three lines?

```python
x = 10       # starting value
x *= 2       # same as x = x * 2
x -= 5       # same as x = x - 5
print(x)
```

Answer: 15. x becomes 20 after `*= 2`, then 15 after `-= 5`. Each line updates the value the line before it left behind.

# 9. Sequence-Controlled vs. Sentinel-Controlled Repetition

### Concept
A sequence-controlled loop repeats a known number of times (`for i in range(n)`). A sentinel-controlled loop repeats until a special stop value shows up in the input.

```python
total = 0.0                          # running sum of readings
for i in range(3):                   # runs exactly 3 times
    reading = float(input(f"Reading {i+1}: "))   # user-entered value
    total += reading
print(f"Average = {total/3:.2f}")    # mean of the 3 readings
```

Detailed notes: Sequence-controlled repetition is the right tool when the number of iterations is known in advance (a fixed batch of sensor readings, a fixed number of trials); sentinel-controlled repetition is right when the end is defined by the data itself (a stop code, an empty line, or a physically meaningful boundary value). The sentinel must be a value real data can never legitimately take (here, -1 for a pressure reading), otherwise a genuine -1 reading would falsely end the loop. The most common sentinel-loop bug, shown in the activity, is using or counting the sentinel value before checking for it.

### Activity: spot the bug
This should stop when the user enters -1, but the final average comes out wrong. What is actually wrong with it?

```python
total, count = 0.0, 0        # running sum and count of readings
while True:                   # loop runs until an explicit break
    reading = float(input("Reading (-1 to stop): "))  # sentinel: -1
    total += reading
    count += 1
    if reading == -1:
        break
print(f"Average = {total/count:.2f}")
```

Answer: -1 gets added to total and counted before the break, so it drags the average down. The check needs to happen right after reading, before `total += reading`.

# 10. Putting It Together: A Stream Monitoring Loop

### Concept
A real monitoring script combines a loop, a conditional, and accumulation together, exactly what the lab exercise will ask you to build.

```python
readings_bar = [1.8, 2.3, 0.9]   # pressure readings, bar
total = 0.0                       # running sum of readings, bar
for P in readings_bar:            # P = current pressure reading, bar
    if P < 1.0:
        status = "low"
    elif P <= 2.0:
        status = "normal"
    else:
        status = "high"
    total += P
print(f"Average P = {total/len(readings_bar):.2f} bar, last status: {status}")
```

Detailed notes: This section introduces no new syntax; it is deliberately a synthesis. A for loop provides repetition, an if/elif/else inside it provides per-item classification, and a running total outside the if provides accumulation. This is the shape of nearly every simple monitoring or data-reduction script students will write in the lab and in later courses: iterate over readings, classify or transform each one, accumulate a summary statistic. Encourage students to name these three ingredients separately when reading any unfamiliar loop-based script; recognizing the pattern is faster than tracing it line by line.

### Activity: cold call
If a fourth reading of 2.5 bar were added, what would the final status print, and would the average go up or down?

```python
readings_bar = [1.8, 2.3, 0.9, 2.5]   # readings, bar (4th added)
total = sum(readings_bar)              # sum of all readings, bar
print(f"Average = {total/len(readings_bar):.2f} bar")
```

Answer: status would be 'high' (2.5 > 2.0), and the average rises from about 1.67 to about 1.88 bar, since 2.5 is above the original average.
