## Part 2: Computers, Data, and How They Fit Together

### 💡 Concept

A computer is **hardware** (physical: CPU, memory, sensors) driven by **software** (instructions you write — Python, Aspen, Excel).

Hardware gets faster and cheaper every year or two — this is **Moore's Law**.

### 💡 Concept — The Data Hierarchy

Bits build up into everything a computer stores:

**Bit → Byte → Field → Record → File → Database → Big Data**

Each level is just a meaningful grouping of the level below it.

### 🔍 Example

A student record: `("Ali Raza", "Chemical Eng.", 3.42)`

- Each **character** (A, l, i, …) is built from **bits**
- The name, major, GPA are each a **field**
- All three fields together = one **record**
- All students' records together = a **file** / **database**

## Part 3: From Machine Code to Python

### 💡 Concept

Programming languages sit on a spectrum: **machine language** (raw 0s/1s) → **assembly** (symbolic shorthand) → **high-level languages** like Python (near-English).

Higher level = easier for humans, translated for you by a compiler/interpreter.

### 💡 Concept — Thinking in Objects

Python is **object-oriented**: data (attributes) and behavior (methods) are bundled into reusable **classes**.

A car's engineering drawing = a *class*. An actual car you drive = an *object* (instance).

### 🔍 Example

Same task, three languages:

```
+1300042774              # machine language
load basepay              # assembly
add overpay
store grosspay
grossPay = basePay + overTimePay   # high-level (Python-like)
```

Pressing the gas pedal = calling a **method** on the car **object**.

### Problem 1 — Reactor Temperature Readings

A reactor recorded temperatures (K) each minute for 8 minutes:

`T = [350, 352, 351, 355, 358, 360, 359, 361]`

**Tasks:** find the average, the minimum, the maximum, and check whether the reactor is warming overall.

### ✍️ Your Turn (30 sec, no laptop) — just eyeball the 8 numbers above: is this reactor warming or cooling overall? Shout out your guess.

````python
# Store the recorded reactor temperatures in a list
T = [350, 352, 351, 355, 358, 360, 359, 361]

# Calculate average temperature (sum divided by number of readings)
avg_T = sum(T) / len(T)

# Find minimum and maximum temperatures in the list
min_T = min(T)
max_T = max(T)

print("Average T (K):", avg_T)
print("Minimum T (K):", min_T)
print("Maximum T (K):", max_T)

# Check if reactor is warming overall: compare last and first reading
warming = T[-1] > T[0]
print("Warming overall?", warming)
````

### 💬 Check

Answer: closer to a record/file — 8 fields (readings) grouped together over one session.

### Problem 2 — Density Calculation

Density = m / V

Suppose m = 500 kg and V = 0.65 m³.

### ✍️ Your Turn (30 sec, no laptop) — is that denser or less dense than water (1000 kg/m³)? Quick guess before we compute it.

### Problem 3 — Temperature Conversions

Convert 80 °C to K and °F.

### ✍️ Your Turn (30 sec, no laptop) — 80 °C: is that closer to freezing or boiling on the Fahrenheit scale? Ballpark it before we run the exact math.

### Problem 4 — Ideal Gas Law

PV = nRT

Given:
- P = 5 bar
- V = 2 m³ = 2000 L
- T = 300 K
- R = 0.08314 bar·L/(mol·K)

Compute moles (n).

### ✍️ Your Turn (30 sec, no laptop) — without doing the math, do you expect this to come out to more or less than 100 moles of gas? Thumbs up or down.

````python
# Define pressure in bar
P_bar = 5.0

# Define volume in cubic meters
V_m3 = 2.0

# Define temperature in Kelvin
T_K = 300.0

# Define gas constant in bar·L/(mol·K)
R_barL = 0.08314

# Convert volume from m³ to liters
V_L = V_m3 * 1000.0

# Calculate number of moles (n = PV/RT)
n = (P_bar * V_L) / (R_barL * T_K)

print("Volume (L):", V_L)
print("Moles of gas:", n)
````

### 💬 Check — we'll come back to this exact formula in Week 5, wrapped up inside a reusable function. Today it's just plugged straight in.

### Problem 5 — Energy Balance: How Much Heat Does It Take?

We want to heat 200 kg of water from 25 °C to 65 °C.

Q = m · Cp · ΔT, where Cp(water) ≈ 4.18 kJ/(kg·K)

### ✍️ Your Turn (30 sec, no laptop) — boiling a kettle for tea takes roughly 200 kJ. Is heating 200 kg of water by 40 °C in the same ballpark, or a different order of magnitude? Guess before we compute.

````python
# Define mass of water in kilograms
m_water = 200.0   # kg

# Define specific heat capacity of water in kJ/(kg*K)
Cp_water = 4.18

# Define temperature change in Kelvin (equal to Celsius for a difference)
delta_T = 65.0 - 25.0

# Calculate heat required (Q = m * Cp * deltaT)
Q = m_water * Cp_water * delta_T

print("Temperature change (K):", delta_T)
print("Heat required (kJ):", Q)
print("Heat required (MJ):", Q / 1000)
````

### 💬 Check — poll: was your order-of-magnitude guess close? This is a simple energy balance — no reaction, no phase change, just sensible heat.

### Mini-Challenge — Mole Fractions (🔭 Sneak Peek: Functions)

Given:
- Mass_A = 36 g, MW_A = 18 g/mol
- Mass_B = 92 g, MW_B = 46 g/mol

Compute mole fractions.

This one wraps the math in a **function** — a reusable "recipe" you'll learn to write yourself in Week 5. For now, just watch what it produces.

### ✍️ Your Turn (30 sec, no laptop) — A has less mass than B, but that doesn't decide the mole fraction. Which substance do you think ends up with the higher mole fraction? Guess.

````python
# Define a function to compute mole fractions for a binary mixture
def mole_fractions(mass_A, mass_B, MW_A, MW_B):
    # Convert masses to moles
    n_A = mass_A / MW_A
    n_B = mass_B / MW_B

    # Calculate total moles
    total = n_A + n_B

    # Mole fractions of A and B
    x_A = n_A / total
    x_B = n_B / total

    return x_A, x_B

# Test the function with given values
xA, xB = mole_fractions(36, 92, 18, 46)

print("x_A:", xA, "x_B:", xB)
````

### 💬 Check — cold-call: in plain words, what does `mole_fractions(...)` let us do that we couldn't easily do by just retyping the formula each time?

Answer: reuse the same calculation for any mass_A/mass_B/MW_A/MW_B without rewriting the math — that's the whole point of a function, previewed here before Week 5.
