# 1. Why Git and GitHub

### Concept
**Git** records the full history of your files on your own machine. **GitHub** hosts that history online so six people can work on one project without overwriting each other.

### Example
```
Working folder --add--> Staging --commit--> Local repo
                                              |
                                            push
                                              v
                                       GitHub (remote)
```

Detailed notes: Git is the version control tool and works fully offline. GitHub is a web service built around Git that adds sharing, issues, pull requests, and access control. A commit is a saved snapshot with an author, a timestamp, and a message; nothing leaves your machine until you push. This three-stage picture (working folder, staging, local repo) explains most beginner confusion.

### Activity: cold call
Your group of six edits one notebook the same night. What goes wrong if you share it by email? What goes wrong in a shared Drive folder?

Answer: email creates final_v2_REAL_final copies and no one knows which is current; a shared folder lets the last save overwrite earlier edits silently. Neither records who changed what or why. Take two or three answers, then name those gaps as exactly what Git fills.

# 2. The Core Cycle: add, commit, push

### Concept
Five commands cover most of your semester: `clone` copies a repo to your machine, `status` shows what changed, `add` stages a file, `commit` saves a snapshot, `push` sends it to GitHub.

### Example
```bash
git clone <url>             # copy the repo to your machine
git status                  # what changed since the last commit?
git add notebooks/01_clean_pressure_data.ipynb   # stage one file
git commit -m "Fix psi to kPa conversion in feed data"   # snapshot
git push                    # send the commit to GitHub
```

Detailed notes: Stage files by name rather than with git add . so you control exactly what enters the commit; a blind add can sweep in large data files, secrets, and checkpoint folders. A good message says what changed and why, in one line. Six weeks later, 'Fix psi to kPa conversion in feed data' is findable; 'update' is not.

### Activity: spot the bug

```

    data/raw/plant_2024.csv
```

Answer: no. Untracked means Git is not watching the file at all. It would need add, commit, and push to reach GitHub, and a raw plant data file may not belong in the repo in the first place (section 4).

# 3. Branches and Pull Requests

### Concept
`main` is the version that always works. A **branch** is your private line of work, and a **pull request** asks the team to merge it into `main` after a teammate reviews it.

### Example
```bash
git switch -c feature/random-forest      # new branch
# edit, add, commit as usual
git push -u origin feature/random-forest   # publish the branch
# then open a Pull Request on GitHub
```

Detailed notes: Working on a branch means a half-finished model never breaks the notebook your teammates run. The pull request is where review happens: a teammate reads the diff, comments, and approves. It also leaves a permanent record of who proposed and who reviewed each change, which is what makes contributions visible. Pull main into your branch often so conflicts stay small.

### Concept
A **merge conflict** appears when two people change the same lines. Git cannot guess who is right, so it marks both versions and you decide.

```
<<<<<<< HEAD
learning_rate = 0.01
=======
learning_rate = 0.001
>>>>>>> feature/tuning
```

Detailed notes: Everything between <<<<<<< and ======= is your branch's version; between ======= and >>>>>>> is the incoming one. Keep one (or combine), delete all three marker lines, then add and commit. Conflicts are normal teamwork, not a sign of failure.

### Activity: think-pair-share
Anna and Bilal both branch from `main` on Monday. Anna merges first on Tuesday, then Bilal edits the same line. Who sees the conflict, and when?

Answer: Bilal, when he tries to merge or update his branch after Anna's merge, because main now holds Anna's version of that line. Listen for students who think Anna sees it. Follow-up: how does pulling main every day shrink the problem?

# 4. Data Science Repositories

### Concept
A project repo has a predictable layout, a `README`, and a `.gitignore`. Large raw data and secrets stay out of it, because GitHub rejects files over 100 MB and history keeps anything you commit.

### Example
```
project/
  data/        small or sample data only
  notebooks/   exploration and analysis
  src/         reusable functions and models
  reports/     figures and final report
  README.md    problem, data source, how to run
  requirements.txt
  .gitignore
```

Detailed notes: This layout is a suggestion; adapt it to your project. The README is the first thing a reader sees, so state the chemical engineering problem, where the data came from and its licence, and the exact steps to run the analysis. requirements.txt lets a teammate recreate your environment. Clear heavy notebook outputs, or Restart and Run All, before committing.

### Concept
A `.gitignore` lists patterns Git must never track. Write it before the first commit.

````python
ignore_patterns = [
    '__pycache__/',         # compiled Python cache
    '.ipynb_checkpoints/',  # Jupyter autosave copies
    '.env',                 # local secrets, such as API keys
    'data/raw/',            # large raw sensor data, kept out of the repo
]
for pattern in ignore_patterns:
    print(pattern)
````

Detailed notes: Each pattern is matched against paths in the repo; a trailing slash means a folder. If a file was already committed, adding it to .gitignore later does not remove it from history, which is why the file is written first. Keep raw data somewhere shared (Drive, lab server) and describe how to obtain it in the README.

### Activity: spot the bug

````python
API_KEY = 'sk-abc123'   # key for the plant data service
print(API_KEY[:3] + '...')
````

Answer: private repos get shared, forked, and leaked, and history keeps the key forever, even after the line is deleted. Read it from the environment instead (os.environ.get('API_KEY')), keep the value in a .env file listed in .gitignore, and rotate any key that was ever committed.

### Activity: predict then run
The limit is 100 MB per file. What does this print for each file size?

````python
limit_MB = 100                                  # GitHub per-file limit, MB
sizes_MB = {'plant_2024.csv': 340, 'summary.csv': 2}   # file sizes, MB
for name, size in sizes_MB.items():
    ok = size < limit_MB                        # True if GitHub accepts it
    print(name, 'OK' if ok else 'too large')
````

Answer: plant_2024.csv too large, summary.csv OK. Link back to the layout: keep the 340 MB raw file out of the repo, commit the small summary, and describe in the README how to obtain the raw data.

# 5. Group Workflow and Assessment

### Concept
Your group works in one private repository in the course organization. Each member uses their own account, tasks are tracked as issues, and changes reach `main` only through reviewed pull requests.

### Example
1. One private repo per group, named `group-NN-short-project-name`.
2. Every member commits under their **own** GitHub account.
3. Each task is an **issue**, assigned to one person.
4. Changes reach `main` only through **pull requests**.
5. A teammate who did not write it reviews every PR.

Detailed notes: Suggested roles for six: data engineer, exploratory analysis and visualization, modelling, machine learning and evaluation, documentation and report, integration lead who reviews and merges. Rotate review duty rather than ownership. Set user.name and user.email in Git to the email registered on GitHub, or commits will not be credited to the right account. Contribution is judged on repository activity, quality of work, peer evaluation, and a short individual walkthrough. Work that never appears in the repo does not count.

### Activity: predict then run
This counts commits per author for one group's history. Who has done the most commits, and does that settle who did the most work?

````python
from collections import Counter

authors = ['Anna', 'Bilal', 'Anna', 'Chen', 'Anna', 'Anna', 'Dina', 'Anna']   # one name per commit
counts = Counter(authors)        # commits per author
print(counts.most_common())
````

Answer: Anna with 5 commits. It does not settle it. Commit counts ignore size and quality, pair programming credits one person, and a wrong Git email hides commits. Ask students what else they would look at: pull requests, reviews, issues closed, and what each commit actually changed. Co-authored-by lines in a commit message credit pair work.

## Recap and Lab Practice
- Git keeps the history on your machine; GitHub shares it with your group.
- Small commits with clear messages; branches and reviewed pull requests protect `main`.
- Keep large data, secrets, and clutter out with a `.gitignore` written first.
- Contribution is judged on what is visible in the repo.
