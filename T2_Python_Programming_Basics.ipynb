# 1. Variables and Assignment

### Concept
A variable is a name that stores a value in memory. Use `=` to assign. Names should be descriptive: `temperature_K`, not `x`.

```python
temperature_K = 350.0    # Kelvin
pressure_bar = 5.0       # bar
flow_mol_s = 2.5         # mol/s

print(temperature_K, pressure_bar, flow_mol_s)
```

*Ties to T-03.01.01/02 Units and Composition (KA-03)*

### Activity: predict then run
`temperature_K` was set above. What happens when we run this?

```python
temperature_K = 365.0
print(temperature_K)
print(Temperature_K)
```

Answer: the second print raises NameError. Python is case-sensitive, so `Temperature_K` was never defined. Reinforces that `flow_mol_s` and `Flow_mol_s` are different names.

# 2. Arithmetic Operators and Precedence

### Concept
`+ - * / ** %` are the core operators. Precedence follows math convention: `**` first, then `* /`, then `+ -`. Use parentheses to make intent explicit.

```python
mass_water = 40.0    # kg
mass_salt = 10.0     # kg

mass_total = mass_water + mass_salt
x_salt = mass_salt / mass_total
print(f"Salt mass fraction = {x_salt:.2f}")
```

*Ties to T-03.01.01/02 Units and Composition (KA-03)*

### Activity: predict then run
Precedence check. What does this print?

```python
print(5 + 3 * 2)
print((5 + 3) * 2)
```

Answer: 11, then 16. Multiplication runs before addition unless parentheses force the addition first.

# 3. Storing Multiple Values: Lists and Tuples

### Concept
A list `[...]` holds an ordered, changeable sequence. A tuple `(...)` is similar but immutable, once created it cannot be changed. Both can be unpacked into named variables.

```python
components = ["Benzene", "Toluene", "Xylene"]
components[1] = "Phenol"     # lists can be edited
print(components)

feed = ("Feed-1", 350.0, 5.0)   # (name, T_K, P_bar)
name, T_K, P_bar = feed
print(name, T_K, P_bar)
```

*Ties to T-03.01.01/02 Units and Composition (KA-03)*

### Activity: spot the bug
This should swap `T_K` and `P_bar` and print both. What is actually wrong with it?

```python
T_K = 350.0
P_bar = 5.0
T_K, P_bar = P_bar
print(T_K, P_bar)
```

Answer: raises TypeError, cannot unpack a single float into two names. The swap needs `T_K, P_bar = P_bar, T_K` (a tuple on the right).

# 4. The Math Module and Unit Conversions

### Concept
`import math` gives access to functions like `math.sin` and constants like `math.pi`. Useful for unit and property conversions beyond basic arithmetic.

```python
import math

T_C = 80.0
T_K = T_C + 273.15
T_F = T_C * 9 / 5 + 32
print(f"{T_C} C -> {T_K} K -> {T_F} F")
```

*Ties to T-03.01.01/02 Units and Composition (KA-03), unit conversion is anchor content for this week*

### Activity: predict then run
`math.sin` takes radians. What does this print, roughly?

```python
print(math.sin(1.2))
print(math.sin(math.pi))
```

Answer: about 0.932, then a value extremely close to 0 (not exactly 0, due to floating point rounding). Good moment to mention floating point is approximate.

# 5. print() and f-string Formatting

### Concept
`print()` joins values with spaces by default. An f-string, `f"...{value:.2f}..."`, formats a number inline, here to 2 decimal places.

```python
mass_kg = 500.0
volume_m3 = 0.65
rho = mass_kg / volume_m3

print("Density =", rho, "kg/m^3")
print(f"Density = {rho:.2f} kg/m^3")
```

### Activity: predict then run
`rho` above is about 769.23. What does each line print?

```python
print(f"{rho:.0f}")
print(f"{rho:.1f}")
print(f"{rho:8.2f}")
```

Answer: 769, then 769.2, then '  769.23' (the 8 sets a minimum field width, padding with spaces on the left).

# 6. Strings and Concatenation

### Concept
Strings are quoted text. Join them with `+`, or prefer an f-string when numbers are involved, it is easier to read.

```python
stream_id = "Feed-1"
T_K = 360.0
P_bar = 3.2

report = f"[{stream_id}] T={T_K:.1f} K, P={P_bar:.1f} bar"
print(report)
```

*Ties to T-03.01.01/02 Units and Composition (KA-03)*

### Activity: spot the bug
This should print `Mixture: Benzene + Phenol (2 parts)`. What is wrong?

```python
comp1 = "Benzene"
comp2 = "Phenol"
n = 2
print("Mixture: " + comp1 + " + " + comp2 + " (" + n + " parts)")
```

Answer: raises TypeError, cannot concatenate str and int with +. Needs str(n), or better, an f-string: f'... ({n} parts)'.

# 7. Getting Input from the User

### Concept
`input()` always returns a string, even if the user types a number. Convert with `float()` or `int()` before doing arithmetic on it.

```python
mass_A = float(input("Mass of water, g: "))
mass_B = float(input("Mass of ethanol, g: "))

n_A, n_B = mass_A / 18.0, mass_B / 46.0
x_A = n_A / (n_A + n_B)
print(f"x_water = {x_A:.3f}")
```

*Ties to T-03.01.01/02 Units and Composition (KA-03)*

### Activity: spot the bug
A student wrote this to read a flow rate. Running it crashes as soon as the user enters a number. Why?

```python
flow_mol_s = input("Enter flow, mol/s: ")
flow_kmol_s = flow_mol_s / 1000
print(flow_kmol_s)
```

Answer: input() returns a string, so flow_mol_s / 1000 raises TypeError. Needs float(input(...)) before the division.

# 8. Putting It Together: A Vessel Report

### Concept
A real script combines variables, arithmetic, and formatted output together, exactly what the lab exercise will ask you to build.

```python
stream_id = "Feed-2"
mass_water, mass_salt = 40.0, 10.0
x_salt = mass_salt / (mass_water + mass_salt)
print(f"[{stream_id}] x_salt={x_salt:.2f}, total={mass_water+mass_salt:.1f} kg")
```

*Ties to T-03.01.01/02 Units and Composition (KA-03)*

### Activity: cold call
If `mass_salt` doubled and `mass_water` stayed the same, what happens to `x_salt`? Explain in one sentence before we check.

```python
mass_salt = 20.0
x_salt = mass_salt / (mass_water + mass_salt)
print(f"x_salt = {x_salt:.2f}")
```

Answer: x_salt rises from 0.20 to 0.33, less than double, since the total mass grows too. Good check for whether students see that a fraction does not scale linearly with one of its inputs.
